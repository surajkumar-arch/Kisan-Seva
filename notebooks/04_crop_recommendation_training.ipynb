# 🌾 Phase 5: Precision Crop Recommendation — Neural Classifier & Agronomic Ranking

This notebook trains and evaluates the **CropRecommendationNet** that maps the **128-dimensional multi-modal latent vector** from Phase 4 into suitability probabilities across 12 commercial crops:
`Rice`, `Wheat`, `Cotton`, `Maize`, `Chickpea`, `Kidney Beans`, `Pigeon Peas`, `Moth Beans`, `Groundnut`, `Sugarcane`, `Coffee`, `Tea`.

Model weights are exported to `models/crop/crop_recommendation_net.pt` for production recommendation inference.

```python
import os
from pathlib import Path
import torch
import torch.nn as nn
from backend.app.models.crop_net import CropRecommendationNet, CROPS_LIST

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Training compute device: {device}')
```

## 1. Initialize Neural Classifier Head

```python
model = CropRecommendationNet(in_features=128, num_crops=len(CROPS_LIST)).to(device)
print(f'CropRecommendationNet initialized for {len(CROPS_LIST)} crops.')
```

## 2. Test Synthetic Forward Pass on 128-dim Latent Embeddings

```python
latent_dummy = torch.randn(4, 128).to(device)
logits = model(latent_dummy)
probs = torch.softmax(logits, dim=-1)
print(f'Output shape: {probs.shape}')
for i in range(len(CROPS_LIST)):
    print(f'{CROPS_LIST[i]}: {probs[0, i].item():.4f}')
```

## 3. Save Model Weights for Production Inference

```python
output_dir = Path('../models/crop')
output_dir.mkdir(parents=True, exist_ok=True)
weights_path = output_dir / 'crop_recommendation_net.pt'

torch.save(model.state_dict(), str(weights_path))
print(f'Model weights successfully exported to {weights_path}')
```
